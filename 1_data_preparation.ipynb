```python
import numpy as np
import pandas as pd

# --- 0) Make sure types + sorting are correct ---
df = df_stocks.copy()

df["Date"] = pd.to_datetime(df["Date"])
df = df.sort_values(["Symbol", "Date"]).reset_index(drop=True)

price_col = "Adj Close"  # use Adj Close for indicators

# --- 1) Returns (direction) ---
g = df.groupby("Symbol", group_keys=False)

df["return_1d"]  = g[price_col].pct_change(1)   # percentage change over 1 day
df["return_5d"]  = g[price_col].pct_change(5)   # percentage change over 5 days
df["return_20d"] = g[price_col].pct_change(20)  # percentage change over 20 days



# --- 2) Moving averages (trend) --- only normalized diffs (stationary, scale-independent) ---
# Raw MA values are non-stationary (grow with price), so we drop them and keep only the % distance
for w in (20, 50, 200):
    ma = g[price_col].transform(lambda x: x.rolling(w, min_periods=w).mean())
    df[f"ma{w}_diff"] = (df[price_col] - ma) / ma  # how far price is from MA, as % of MA



# --- 3) Volatility (risk) ---
# Rolling stddev of daily returns over 20 days — whether a stock is stable or volatile recently
df["volatility_20"] = g["return_1d"].transform(lambda x: x.rolling(20, min_periods=20).std())



# --- 4) RSI (14) --- Relative Strength Index
# "How strong have recent gains been relative to recent losses?"
# 0 to 100 scale — High: recent gains dominate, Low: recent losses dominate
"""
RSI value    Interpretation
70-100       Overbought / strong upward momentum
50-70        Healthy uptrend
~50          Balanced (no strong momentum)
30-50        Downtrend
0-30         Oversold / strong downward momentum
"""

def rsi_14(series: pd.Series, window: int = 14) -> pd.Series:
    delta = series.diff()
    gain = delta.clip(lower=0)
    loss = -delta.clip(upper=0)

    # Wilder-style smoothing (EMA with alpha=1/window)
    avg_gain = gain.ewm(alpha=1/window, adjust=False, min_periods=window).mean()
    avg_loss = loss.ewm(alpha=1/window, adjust=False, min_periods=window).mean()

    rs = avg_gain / (avg_loss + 1e-12)  # avoid division by zero
    return 100 - (100 / (1 + rs))

df["rsi_14"] = g[price_col].transform(rsi_14)



# --- 5) Range + ATR (14) ---
df["range_pct"] = (df["High"] - df["Low"]) / df[price_col]  # intraday move relative to price

# True Range components (need prev close per symbol)
prev_close = g[price_col].shift(1)
tr1 = (df["High"] - df["Low"]).abs()        # intraday range
tr2 = (df["High"] - prev_close).abs()       # gap up risk
tr3 = (df["Low"]  - prev_close).abs()       # gap down risk
true_range = np.maximum(tr1, np.maximum(tr2, tr3))

atr_14 = true_range.groupby(df["Symbol"]).transform(lambda x: x.rolling(14, min_periods=14).mean())
df["atr_pct"] = atr_14 / df[price_col]  # ATR as % of price (stationary, scale-independent)
# Note: raw atr_14 (absolute $) is dropped — it's price-scale dependent and non-stationary



# --- 6) Volume features ---
vol_avg_20 = g["Volume"].transform(lambda x: x.rolling(20, min_periods=20).mean())
vol_std_20 = g["Volume"].transform(lambda x: x.rolling(20, min_periods=20).std())

df["vol_ratio"] = df["Volume"] / (vol_avg_20 + 1e-12)             # today's volume vs 20d average
df["vol_z_20"]  = (df["Volume"] - vol_avg_20) / (vol_std_20 + 1e-12)  # how unusual today's volume is



# --- 7) Drop rows where indicators are not yet available (early history per symbol) ---
feature_cols = [
    "return_1d", "return_5d", "return_20d",   # momentum
    "ma20_diff", "ma50_diff", "ma200_diff",   # trend (stationary, normalized)
    "volatility_20",                           # risk
    "rsi_14",                                  # overbought/oversold
    "range_pct", "atr_pct",                   # intraday volatility (normalized)
    "vol_ratio", "vol_z_20",                  # volume signals
]
df_stocks_transformed = df.dropna(subset=feature_cols).reset_index(drop=True)

df_stocks_transformed
```
